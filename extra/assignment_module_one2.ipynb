````python
import cv2
import numpy as np
from pathlib import Path
import time
import matplotlib.pyplot as plt

# ============================================================
# CONFIG
# ============================================================

TARGET_DIR = "coin_dataset/target_set3"
TEMPLATE_DIR = "coin_dataset/reference_set"

COIN_VALUES = {
    "1cent": 0.01,
    "2cent": 0.02,
    "5cent": 0.05,
    "10cent": 0.10,
    "20cent": 0.20,
    "50cent": 0.50,
    "1euro": 1.00,
    "2euro": 2.00
}

# ============================================================
# SIFT + FLANN
# ============================================================

sift = cv2.SIFT_create(
    nfeatures=2000,
    contrastThreshold=0.01,
    edgeThreshold=10,
    sigma=1.2

)

FLANN_INDEX_KDTREE = 1
index_params = dict(algorithm=FLANN_INDEX_KDTREE, trees=5)
search_params = dict(checks=50)

flann = cv2.FlannBasedMatcher(index_params, search_params)

# ============================================================
# TEMPLATE LOADING
# ============================================================

def normalize_coin_name(name):
    return name.lower().replace("_", "").replace(" ", "")



# ============================================================
# ROI
# ============================================================


def extract_roi(img, x, y, r):

    pad = int(r)

    try:
        x1 = max(0, x - pad)
        x2 = min(img.shape[1], x + pad)

        y1 = max(0, y - pad)
        y2 = min(img.shape[0], y + pad)

    except:
        return None

    roi = img[y1:y2, x1:x2]

    if roi is None or roi.size == 0:
        return None

    # ============================================================
    # CREATE CIRCULAR MASK
    # ============================================================

    mask = np.zeros(roi.shape[:2], dtype=np.uint8)

    center_x = roi.shape[1] // 2
    center_y = roi.shape[0] // 2

    radius = int(min(center_x, center_y) * 0.95)

    cv2.circle(
        mask,
        (center_x, center_y),
        radius,
        255,
        -1
    )

    # ============================================================
    # REMOVE BACKGROUND
    # ============================================================

    masked_roi = cv2.bitwise_and(
        roi,
        roi,
        mask=mask
    )

    ##====================================================================================================================================================================================
    # return mask too

    return masked_roi


def gaussian(img_gray, name):

    # ============================================================
    # gaussian filter
    # ============================================================

    kernel = np.array([
        [-1,-1,-1],
        [-1, 9,-1],
        [-1,-1,-1]
    ])
            
    # Calculate k_size
    sigma = 1.5
    k_size = int(np.ceil((3*sigma))*2 + 1)  # rule of thumb for a good kernel size given sigma

    # Find the 1D Gaussian kernel
    gaussian_kernel_1D = cv2.getGaussianKernel(k_size, sigma)
    

    # Apply the kernel
    start = time.time()
    print(f"\nApplying Gaussian filter with sigma={sigma} and k_size={k_size}...", gaussian_kernel_1D)
    # img_tr = cv2.filter2D(img_gray, -1, kernel)
    img_tr = cv2.filter2D(img_gray, -1, gaussian_kernel_1D)
    # img_tr = cv2.filter2D(img_tr, -1, gaussian_kernel_1D.transpose())
    end = time.time()
    # print(f"\nTime needed to apply a 1D convolution twice: {(end - start):.3f} seconds.")

    # show_img(img_tr, f"Gaussian 1D for {name}")

    # Find the 2D Gaussian kernel
    gaussian_kernel_2D = gaussian_kernel_1D.dot(gaussian_kernel_1D.transpose())

    # Apply the kernel
    start = time.time()
    img_tr = cv2.filter2D(img_gray, -1, gaussian_kernel_2D)
    end = time.time()
    # print(f"\nTime needed to apply a single 2D convolution: {(end - start):.3f} seconds.")

    # img_gray = img_tr  # update img_gray for next steps

    # show_img(img_tr, f"Gaussian 2D for {name}")



    # # Use the OpenCV function GaussianBlur
    # start = time.time()
    # img_tr = cv2.GaussianBlur(img_gray, (k_size, k_size), sigma)
    # end = time.time()
    # print(f"\nTime needed to apply the OpenCV API: {(end - start):.3f} seconds.")

    # # Display the image after filtering
    # plt.imshow(img_tr, cmap="gray", vmin=0, vmax=255)
    # plt.title(f"[TEMPLATE GAUSSIAN OpenCV] {name}")
    # plt.axis("off")
    # plt.show()

    return img_tr


def mean_filter(img_gray, name):
    
    # ============================================================
    # mean filter
    # ============================================================
    
    k_size = 9
    mean_kernel = np.ones([k_size, k_size])/(k_size**2)

    img_tr = cv2.filter2D(img_gray, -1, mean_kernel)


    return img_tr


def median_filter(img_gray, name):
    
    # ============================================================
    # median filter
    # ============================================================
    
    k_size = 9
    img_tr = cv2.medianBlur(img_gray, k_size)

    return img_tr


def bilateral_filter(img_gray, name):   
    
    # ============================================================
    # bilateral filter
    # ============================================================
    
    img_tr = cv2.bilateralFilter(img_gray, d=9, sigmaColor=75, sigmaSpace=75)


    return img_tr


def sopel_filter(img_gray, name):
    # Define sobel kernels
    sobel_kernel_x = np.array([[-1,0,1],[-2,0,2],[-1,0,1]]) * 1/4
    sobel_kernel_y = np.array([[-1,-2,-1],[0,0,0],[1,2,1]]) * 1/4

    # Compute |dI(x,y)/dx|
    dx = cv2.filter2D(img_gray.astype(float), -1, sobel_kernel_x)
    dx = np.abs(dx)

    # Display dx
    plt.imshow(dx, cmap="gray", vmin=0, vmax=255)
    plt.show()

    # Compute |dI(x,y)/dy|
    dy = cv2.filter2D(img_gray.astype(float), -1, sobel_kernel_y)
    dy = np.abs(dy)

    # Display dy
    plt.imshow(dy, cmap="gray", vmin=0, vmax=255)
    plt.show()

    # Compute the pixel-wise gradient module
    sobel = np.maximum(dx, dy)

    # Display the gradient module
    plt.imshow(sobel, cmap="gray", vmin=0, vmax=255)
    plt.show()

    return sobel


def laplacian_filter(img_gray, name):
    # Define Laplacian kernel
    laplacian_kernel = np.array([[0,1,0],[1,-4,1],[0,1,0]])

    # Compute the Laplacian response
    laplacian = cv2.filter2D(img_gray.astype(float), -1, laplacian_kernel)

    return laplacian

def canny(img_gray, name):
    
    edges = cv2.Canny(img_gray, 80, 160)

    return edges


def show_img(img, name, debug=True):
    plt.imshow(img, cmap="gray", vmin=0, vmax=255)
    plt.title(name)
    # plt.axis("off")
    plt.show()


def pipeline_template(img_color, name):
    img_gray = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)


    # img_gray = gaussian(img_gray, name)
    # img_gray = mean_filter(img_gray, name)
    # img_gray = median_filter(img_gray, name)
    # img_gray = bilateral_filter(img_gray, name)
    # img_gray = sopel_filter(img_gray, name)
    # img_gray = laplacian_filter(img_gray, name)
    # img_gray = canny(img_gray, name)



    # ============================================================
    # HOUGH CIRCLES (USE GRAYSCALE ONLY)
    # ============================================================
    circles = cv2.HoughCircles(
        img_gray,
        # edges,
        cv2.HOUGH_GRADIENT,
        dp=1.5,
        minDist=80,
        param1=120,
        param2=30,
        minRadius=50,
        maxRadius=140
    )



    circles = np.uint16(np.around(circles[0]))
    x, y, r = circles[0]


    # print('before roi:', img_color.shape)
    # # show the circle
    # debug_img = img_color.copy()
    # cv2.circle(debug_img, (x, y), r, (255, 0, 0), 2)  # blue circle
    # plt.figure(figsize=(6, 6))
    # plt.imshow(cv2.cvtColor(debug_img, cv2.COLOR_BGR2RGB))
    # plt.title(f"Hough Circle for {name}")
    # plt.axis("off")

    # ============================================================
    # ROI FROM COLOR IMAGE (IMPORTANT FIX)
    # ============================================================
    roi = extract_roi(img_color, x, y, r)



    # ============================================================
    # Upsampeling image
    # ============================================================
    # show_img(roi, f"Original ROI for {name}")

    print(f"Original ROI size: {roi.shape[1]}x{roi.shape[0]}")
    roi = cv2.resize(roi, (256, 256), interpolation=cv2.INTER_CUBIC)
    print(f"Upsampled ROI size: {roi.shape[1]}x{roi.shape[0]}")

    # show_img(roi, f"Upsampling for {name}")

    return roi


def load_templates(debug=True):
    templates = {}

    for path in Path(TEMPLATE_DIR).glob("*.jpg"):
        name = normalize_coin_name(path.stem)

        # ============================================================
        # LOAD IMAGE (KEEP COLOR + GRAY)
        # ============================================================
        img_color = cv2.imread(str(path))  # BGR (KEEP THIS)
        


        roi = pipeline_template(img_color, name)
        # ============================================================
        # SIFT (CONVERT ROI TO GRAY ONLY FOR FEATURES)
        # ============================================================
           # ============================================================   # ============================================================   # ============================================================
        sift = cv2.SIFT_create(
            nfeatures=2000,
            contrastThreshold=0.01,
            edgeThreshold=10,
            sigma=1.2
        )
        roi_gray = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)

        # kp = sift.detect(roi_gray, None)

        clahe = cv2.createCLAHE(
            clipLimit=3.0,
            tileGridSize=(8,8)
        )

        roi_gray = clahe.apply(roi_gray)


        kp, des = sift.detectAndCompute(roi_gray, None)

        img_visualization = cv2.drawKeypoints(
            roi_gray,
            kp,
            None,
            flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
        )

        show_img(img_visualization,  f"SIFT Keypoints for {name}")

        print("num keypoints:", len(kp))

        templates[name] = (kp, des)  # store kp only for now, des will be computed later if needed


        # ============================================================
        # BRISK (ADDED ALTERNATIVE FEATURE DETECTOR)
        # ============================================================
        # brisk = cv2.BRISK_create()
        # kp_brisk = brisk.detect(roi_gray, None)

        # img_visualization_brisk = cv2.drawKeypoints(
        #     roi_gray,
        #     kp_brisk,
        #     None,
        #     flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
        # )

        # plt.imshow(img_visualization_brisk)
        # plt.title(f"[TEMPLATE BRISK KEYPOINTS] {name}")
        # # plt.axis("off")
        # plt.show()
        # print("num keypoints (BRISK):", len(kp_brisk))


        # # ============================================================
        # # ORB (ADDED ALTERNATIVE FEATURE DETECTOR)
        # # ============================================================
        # orb = cv2.ORB_create()
        # kp_orb = orb.detect(roi_gray, None)

        # img_visualization_orb = cv2.drawKeypoints(
        #     roi_gray,
        #     kp_orb,
        #     None,
        #     flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
        # )

        # plt.imshow(img_visualization_orb)
        # plt.title(f"[TEMPLATE ORB KEYPOINTS] {name}")
        # # plt.axis("off")
        # plt.show()
        # print("num keypoints (ORB):", len(kp_orb))


    return templates

templates_db = load_templates()




# ============================================================
# CIRCLE DETECTION (UNCHANGED LOGIC)
# ============================================================

# def detect_circles(gray):

#     blur = cv2.GaussianBlur(gray, (9, 9), 2)
#     edges = cv2.Canny(blur, 80, 160)

#     circles = cv2.HoughCircles(
#         # edges,
#         blur,
#         cv2.HOUGH_GRADIENT,
#         dp=1.5,
#         minDist=80,
#         param1=120,
#         param2=30,
#         minRadius=50,
#         maxRadius=140
#     )

#     # debug_img = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

#     # if circles is not None:
#     #     print("Raw circles detected:", len(circles[0]))

#     #     for (x, y, r) in np.uint16(np.around(circles[0])):
#     #         cv2.circle(debug_img, (x, y), r, (255, 0, 0), 2)  # blue circles
#     # else:
#     #     print("Raw circles detected: 0")

#     # plt.figure(figsize=(6, 6))
#     # plt.imshow(cv2.cvtColor(debug_img, cv2.COLOR_BGR2RGB))
#     # plt.title("RAW Hough Circles (Before Filtering)")
#     # plt.axis("off")
#     # plt.show()


#     if circles is None:
#         return []

#     circles = np.uint16(np.around(circles[0]))

#     filtered = []
#     print(f"Detected {len(circles)} circles, filtering...")
    
#     for (x, y, r) in circles:

#         mask = np.zeros(gray.shape, dtype=np.uint8)
#         cv2.circle(mask, (x, y), r, 255, 2)
#         # cv2.circle(mask, (x, y), r, 255, 2)

#         score = np.sum(edges & mask) / (np.sum(mask > 0) + 1)
#         print(f"Circle at ({x}, {y}), r={r}: score={score:.4f}")
#         if score < 0.25:
#             continue

#         filtered.append((x, y, r))

#     return filtered

# ============================================================
# CLASSIFIER (UNCHANGED)
# ============================================================

# def classify_coin(roi_gray):

#     # ============================================================
#     # STEP 1: SIFT on target coin
#     # ============================================================
#     # gray = cv2.cvtColor(roi_gray, cv2.COLOR_BGR2GRAY)

#     # clahe = cv2.createCLAHE(
#     # clipLimit=2.0,
#     #     tileGridSize=(8,8)
#     # )

#     # roi_gray = clahe.apply(roi_gray)

#     # kp1, des1 = sift.detectAndCompute( roi_gray, None)

#     #  STEP 1: convert to grayscale properly
#     gray = cv2.cvtColor(roi_gray, cv2.COLOR_BGR2GRAY)

#     # ensure correct type
#     gray = gray.astype(np.uint8)

#     # CLAHE
#     clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
#     gray = clahe.apply(gray)

#     # SIFT on processed image
#     kp1, des1 = sift.detectAndCompute(gray, None)


#     img_with_keypoints = cv2.drawKeypoints(
#         roi_gray,
#         kp1,
#         None,
#         (0,255,0),
#         cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
#     )

#     show_img(img_with_keypoints, "SIFT Keypoints on Target ROI")

#     if des1 is None or len(des1) < 4:
#         return None

#     des1 = np.float32(des1)

#     best = None
#     best_score = 0

#     # ============================================================
#     # STEP 2: compare with templates
#     # ============================================================

#     # for name, (template_img, kp2, des2) in templates_db.items():

#     print("\nComparing with templates...", templates_db.keys())
#     print("Number of templates:", len(templates_db))    
#     print("Number of keypoints in target ROI:", len(kp1))

#     for name, (kp2, des2) in templates_db.items():

#         if des2 is None:
#             continue

#         des2 = np.float32(des2)

#         try:
#             matches = flann.knnMatch(
#                 des1,
#                 des2,
#                 k=2
#             )

#         except:
#             continue

#         # ============================================================
#         # Lowe ratio test
#         # ============================================================

#         good = []

#         for m_n in matches:

#             if len(m_n) != 2:
#                 continue

#             m, n = m_n

#             # if m.distance < 0.75 * n.distance:
#             if m.distance < 0.82 * n.distance:
#                 good.append(m)

#         # score = len(good) / (len(matches) + 1e-6)

#         # confidence = len(good) / (len(des1) + 1e-6)

#         # final_score = score * confidence



#         score = len(good) / (len(matches) + 1e-6)
#         confidence = len(good) / (len(des1) + 1e-6)
#         final_score = score * confidence




#         # # Option A — ratio of good matches to target descriptors (simpler, cleaner)
#         # final_score = len(good) / (len(des1) + 1e-6)

#         # # Option B — also normalize by template size (good for unbalanced kp counts)
#         # final_score = len(good) / (min(len(des1), len(des2)) + 1e-6)

#         # ============================================================
#         # DEBUG PRINT
#         # ============================================================

#         print("\n----------------")
#         print("Template:", name)

#         print(
#             "Target kp:",
#             len(des1)
#         )

#         print(
#             "Template kp:",
#             len(des2)
#         )

#         print(
#             "Raw matches:",
#             len(matches)
#         )

#         print(
#             "Good matches:",
#             len(good)
#         )

#         # print(
#         #     "Score:",
#         #     round(score,5)
#         # )

#         # print(
#         #     "Confidence:",
#         #     round(confidence,5)
#         # )

#         print(
#             "FINAL:",
#             round(final_score,5)
#         )

#         # ============================================================
#         # keep best match
#         # ============================================================

#         if final_score > best_score:

#             best_score = final_score
#             best = name

#     print(
#         "\nBEST:",
#         best,
#         " SCORE:",
#         best_score
#     )

#     if best_score < 0.002:
#         return None

#     return best


def classify_coin(roi_bgr):
    """Returns (coin_name, score) or (None, 0)."""

    # ── 1. Prepare query ──────────────────────────────────────────
    gray = cv2.cvtColor(roi_bgr, cv2.COLOR_BGR2GRAY).astype(np.uint8)
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
    gray = clahe.apply(gray)

    kp1, des1 = sift.detectAndCompute(gray, None)

    if des1 is None or len(des1) < 8:
        return None, 0

    des1 = np.float32(des1)

    # ── 2. Color histogram for the query coin ────────────────────
    hsv_query = cv2.cvtColor(roi_bgr, cv2.COLOR_BGR2HSV)

    best_name  = None
    best_score = 0.0

    for name, template_data in templates_db.items():
        kp2, des2, roi_template_bgr = template_data   # ← store ROI in load_templates too

        if des2 is None or len(des2) < 8:
            continue

        des2 = np.float32(des2)

        # ── 3. Color similarity gate (fast pre-filter) ───────────
        hsv_tmpl = cv2.cvtColor(roi_template_bgr, cv2.COLOR_BGR2HSV)
        hist_q = cv2.calcHist([hsv_query], [0, 1], None, [36, 32],
                              [0, 180, 0, 256])
        hist_t = cv2.calcHist([hsv_tmpl],  [0, 1], None, [36, 32],
                              [0, 180, 0, 256])
        cv2.normalize(hist_q, hist_q)
        cv2.normalize(hist_t, hist_t)
        color_sim = cv2.compareHist(hist_q, hist_t, cv2.HISTCMP_CORREL)

        if color_sim < 0.30:          # skip clearly wrong coin family
            continue

        # ── 4. FLANN matching ─────────────────────────────────────
        try:
            matches = flann.knnMatch(des1, des2, k=2)
        except Exception:
            continue

        # ── 5. Lowe ratio test (tighter = 0.75) ───────────────────
        good = [m for m, n in matches
                if len((m, n)) == 2 and m.distance < 0.75 * n.distance]

        if len(good) < 8:
            continue

        # ── 6. RANSAC geometric verification ─────────────────────
        src_pts = np.float32([kp1[m.queryIdx].pt for m in good]).reshape(-1,1,2)
        dst_pts = np.float32([kp2[m.trainIdx].pt for m in good]).reshape(-1,1,2)

        _, mask = cv2.findHomography(src_pts, dst_pts,
                                     cv2.RANSAC, ransacReprojThreshold=5.0)

        if mask is None:
            inliers = 0
        else:
            inliers = int(mask.sum())

        if inliers < 6:
            continue

        # ── 7. Score: inliers / min(kp_query, kp_template) ───────
        #    Single ratio — no squaring, properly bounded [0,1]
        score = inliers / (min(len(des1), len(des2)) + 1e-6)

        # Optionally blend color similarity in
        # score = 0.7 * score + 0.3 * max(0, color_sim)

        print(f"  {name:10s}  inliers={inliers:3d}  color={color_sim:.2f}  score={score:.5f}")

        if score > best_score:
            best_score = score
            best_name  = name

    print(f"\n  BEST: {best_name}  SCORE: {best_score:.5f}")
    return (best_name if best_score > 0.005 else None), best_score
    
# ============================================================
# PROCESS IMAGE (WITH DEBUG VISUALS ADDED)
# ============================================================

def process_image(img_path, debug=False):

    img = cv2.imread(str(img_path))
    if img is None:
        return [], 0, None

    # ============================================================
    # STEP 1: SAME PREPROCESSING AS TEMPLATE (for HoughCircles)
    # ============================================================

    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    # blur = cv2.GaussianBlur(gray, (9, 9), 2)

    # img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    blur = gaussian(img_gray, img_path.stem)

    # ============================================================
    # STEP 2: CIRCLE DETECTION
    # ============================================================

    circles = cv2.HoughCircles(
        blur,
        cv2.HOUGH_GRADIENT,
        dp=1.5,
        minDist=80,
        param1=120,
        param2=65,
        minRadius=50,
        maxRadius=140
    )

    
    # show inmage with detected circles for debug
    debug_img = img.copy()
    if circles is not None:
        for (x, y, r) in np.uint16(np.around(circles[0])):
            cv2.circle(debug_img, (x, y), r, (255, 0, 0), 2)  # blue circles
    plt.figure(figsize=(6, 6))
    plt.imshow(cv2.cvtColor(debug_img, cv2.COLOR_BGR2RGB))
    plt.title(f"Raw Hough Circles for {img_path.stem}")
    plt.axis("off")
    plt.show()
    print(f"Circles detected: {len(circles[0]) if circles is not None else 0}")
    

    detected = []
    total = 0

    vis = img.copy()

    if circles is None:
        return [], 0, vis

    circles = np.uint16(np.around(circles[0]))


    


    # ============================================================
    # STEP 3: PROCESS EACH CIRCLE
    # ============================================================

    for (x, y, r) in circles:

        # ROI extraction (same as template)
        roi = extract_roi(img, x, y, r)

        if roi is None:
            continue

        # # preprocessing (same as template)
        # roi_gray = preprocess(roi, debug=False)

        # if roi_gray is None:
        #     continue


        show_img(roi, f"Original ROI for {img_path.stem}")

        print(f"Original ROI size: {roi.shape[1]}x{roi.shape[0]}")
        roi = cv2.resize(roi, (256, 256), interpolation=cv2.INTER_CUBIC)
        print(f"Upsampled ROI size: {roi.shape[1]}x{roi.shape[0]}")

        show_img(roi, f"Upsampling for {img_path.stem}")


        # classification
        coin = classify_coin(roi)

        if coin is None:
            continue

        value = COIN_VALUES.get(coin, 0)

        detected.append((coin, value))
        total += value

        # ============================================================
        # VISUALIZATION (same style as template debug)
        # ============================================================

        cv2.circle(vis, (x, y), r, (0, 255, 0), 2)
        cv2.putText(
            vis,
            coin,
            (x - 20, y),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (0, 255, 0),
            1
        )

    return detected, total, vis

# ============================================================
# RUN
# ============================================================

grand_total = 0

for img_path in Path(TARGET_DIR).glob("*.jpg"):

    coins, total, vis = process_image(img_path, debug=True)

    print(f"\n{img_path.name}")

    for c, v in coins:
        print(f"  {c}: {v}")

    print(f"  Partial: {total:.2f} €")

    grand_total += total

    plt.figure(figsize=(6,6))
    plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
    plt.title("Detection")
    plt.axis("off")
    plt.show()

print("\n====================")
print(f"TOTAL: {grand_total:.2f} €")

````
